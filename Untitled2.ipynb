```python
import pandas as pd
import numpy as np

# Load the dataset using the correct absolute path
data = pd.read_csv("/content/02_student_performance.csv")

# 1. Display dataset information
print("Number of rows and columns:")
print(data.shape)

print("\nColumn names:")
print(data.columns)

print("\nData types:")
print(data.dtypes)


# 2. Check missing values and duplicate Student_ID
print("\nMissing values:")
print(data.isnull().sum())

print("\nDuplicate Student IDs:")
print(data["Student_ID"].duplicated().sum())


# 3. Fill missing attendance with average attendance
data["Attendance_Percentage"] = data["Attendance_Percentage"].fillna(
    data["Attendance_Percentage"].mean()
)


# 4. Check marks outside 0-100
print("\nChecking marks:")

for column in ["Python_Marks", "Math_Marks", "English_Marks"]:
    for mark in data[column]:
        if mark < 0 or mark > 100:
            print(column, mark)


# 5. Calculate total and average marks
data["Total_Marks"] = (
    data["Python_Marks"]
    + data["Math_Marks"]
    + data["English_Marks"]
)

data["Average_Marks"] = data["Total_Marks"] / 3

print("\nStudent marks:")
print(data[["Student_ID", "Total_Marks", "Average_Marks"]])


# 6. Create Grade column
def grade(marks):

    if marks >= 80:
        return "A"

    elif marks >= 60:
        return "B"

    elif marks >= 40:
        return "C"

    else:
        return "F"


data["Grade"] = data["Average_Marks"].apply(grade)

print("\nGrades:")
print(data[["Student_ID", "Average_Marks", "Grade"]])


# 7. Department-wise average marks
print("\nDepartment-wise Average Marks:")
print(
    data.groupby("Department")["Average_Marks"].mean()
)


# 8. Top 10 students
print("\nTop 10 Students:")

top_students = data.sort_values(
    "Average_Marks",
    ascending=False
)

print(
    top_students[
        ["Student_ID", "Average_Marks"]
    ].head(10)
)


# 9. Students below 40%
print("\nStudents below 40%:")

low_students = data[
    data["Average_Marks"] < 40
]

print(
    low_students[
        ["Student_ID", "Average_Marks"]
    ]
)


# 10. Compare attendance bands
def attendance_band(attendance):

    if attendance < 60:
        return "Low"

    elif attendance < 80:
        return "Medium"

    else:
        return "High"


data["Attendance_Band"] = data["Attendance_Percentage"].apply(
    attendance_band
)

print("\nAverage marks by attendance:")
print(
    data.groupby("Attendance_Band")["Average_Marks"].mean()
)


# 11. Study hours and average marks
print("\nStudy Hours and Average Marks:")

print(
    data[
        ["Study_Hours_Per_Week", "Average_Marks"]
    ].sort_values("Study_Hours_Per_Week")
)


# 12. Internet access comparison
print("\nAverage marks with Internet Access:")
print(
    data.groupby("Internet_Access")["Average_Marks"].mean()
)


# 13. NumPy calculations
python_marks = data["Python_Marks"].values

print("\nPython Marks Statistics:")

print("Mean =", np.mean(python_marks))
print("Median =", np.median(python_marks))
print("Standard Deviation =", np.std(python_marks))

Q1 = np.percentile(python_marks, 25)
Q3 = np.percentile(python_marks, 75)

print("Q1 =", Q1)
print("Q3 =", Q3)


# 14. IQR outliers
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("\nPython Marks Outliers:")

for mark in python_marks:

    if mark < lower or mark > upper:
        print(mark)


# 15. Recommendations
print("\nFive Recommendations:")

print("1. Students with low marks should receive extra classes.")
print("2. Students should maintain regular attendance.")
print("3. Students should increase their study hours.")
print("4. Teachers should monitor students scoring below 40%.")
print("5. Students should use online study resources when needed.")
```
